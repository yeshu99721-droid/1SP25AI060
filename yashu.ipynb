```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

iris = load_iris()
X = iris.data

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X)

true_labels = iris.target

df = pd.DataFrame(X, columns=iris.feature_names)
df['cluster'] = clusters
df['True Label'] = true_labels

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=clusters,
    palette='Set2',
    s=60
)

plt.title("K-Means Clustering (k=3) on Iris Dataset")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend(title="Cluster")
plt.show()
```

```python
%pip install MiniSom

from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler
from minisom import MiniSom
import matplotlib.pyplot as plt
import numpy as np

iris = load_iris()
X = iris.data

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

som = MiniSom(
    x=7,
    y=7,
    input_len=4,
    sigma=1.0,
    learning_rate=0.5,
    random_seed=42
)

som.random_weights_init(X_scaled)

som.train_random(X_scaled, num_iteration=100)

plt.figure(figsize=(8, 8))

plt.pcolor(
    som.distance_map().T,
    cmap="coolwarm"
)

plt.colorbar(label="Distance")
plt.title("SOM - Iris Dataset (U-Matrix)")
plt.show()

winner_coordinates = np.array([
    som.winner(x) for x in X_scaled
])

clusters = np.ravel_multi_index(
    winner_coordinates.T,
    dims=(7, 7)
)

print("Cluster assignments:")
print(clusters)
```

```python
import random
import math
def generate_uniform():
    return random.uniform(0,1)
def f(x):
    return math.sin(math.pi*x)
samples = [generate_uniform() for _ in range(10)]
f_values=[f(x) for x in samples]
print("Function values f(x):",f_values)
I_monte_carlo=sum(f_values)/len(f_values)
print("estimated integral(monte carlo):",I_monte_carlo)
```
